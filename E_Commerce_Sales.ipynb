```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries loaded!")
```

```python
df1 = pd.read_csv('/content/query1.csv', header=None,
                   names=['Region', 'Total_Sales', 'Total_Profit', 'Total_Orders'])
print(df1)
```

```python
plt.figure(figsize=(8,5))
plt.bar(df1['Region'], df1['Total_Sales'], color='steelblue')
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.savefig('region_sales_chart.png')
plt.show()
```

```python
df7 = pd.read_csv('/content/query7.csv', header=None,
                   names=['Order_Year', 'Order_Month', 'Total_Sales', 'Total_Profit'])
print(df7)
```

```python
df7['Period'] = df7['Order_Year'].astype(str) + '-' + df7['Order_Month'].astype(str)

plt.figure(figsize=(12,5))
plt.plot(df7['Period'], df7['Total_Sales'], marker='o', color='green')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)
plt.savefig('monthly_sales_trend.png')
plt.show()
```

```python
df2 = pd.read_csv('/content/query2.csv', header=None,
                   names=['Category', 'Total_Sales', 'Total_Profit', 'Total_Orders'])
print(df2)
```

```python
plt.figure(figsize=(7,7))
plt.pie(df2['Total_Sales'], labels=df2['Category'], autopct='%1.1f%%',
        colors=['#4C72B0', '#DD8452', '#55A868'])
plt.title('Sales Share by Category')
plt.savefig('category_sales_pie.png')
plt.show()
```

```python
df6 = pd.read_csv('/content/query6.csv', header=None,
                   names=['Sub_Category', 'Total_Sales', 'Total_Profit'])
print(df6)
```

```python
plt.figure(figsize=(8,5))
plt.barh(df6['Sub_Category'], df6['Total_Profit'], color='crimson')
plt.title('Loss-Making Sub-Categories')
plt.xlabel('Total Profit (Loss)')
plt.ylabel('Sub-Category')
plt.savefig('loss_subcategories.png')
plt.show()
```

```python
df12 = pd.read_csv('/content/query12.csv', header=None,
                    names=['Region', 'Product_Name', 'Total_Sales', 'Sales_Rank'])
print(df12)
```

```python
plt.figure(figsize=(10,6))
plt.bar(df12['Region'], df12['Total_Sales'], color='purple')
plt.title('Top-Selling Product Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
```

```python
plt.figure(figsize=(10,6))
plt.bar(df12['Region'], df12['Total_Sales'], color='purple')
plt.title('Top-Selling Product Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')

for i, row in df12.iterrows():
    plt.text(i, row['Total_Sales'], row['Product_Name'], ha='center', va='bottom', fontsize=8, rotation=15)

plt.savefig('region_top_product.png')
plt.show()
```

```python
df15 = pd.read_csv('/content/query15.csv', header=None,
                    names=['Order_Year', 'Yearly_Sales', 'Prev_Year_Sales', 'YoY_Growth_Percent'])
print(df15)
```

```python
plt.figure(figsize=(8,5))
colors = ['green' if x >= 0 else 'red' for x in df15['YoY_Growth_Percent']]
plt.bar(df15['Order_Year'].astype(str), df15['YoY_Growth_Percent'], color=colors)
plt.title('Year-over-Year Sales Growth %')
plt.xlabel('Year')
plt.ylabel('Growth %')
plt.axhline(0, color='black', linewidth=0.8)
plt.savefig('yoy_growth.png')
plt.show()
```
